```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.cm import ScalarMappable
from matplotlib.colors import Normalize

# Load the dataset
df = pd.read_csv("/run/sample data/earthquake_1995-2023.csv")

# Display first 5 records
print("First 5 records:")
print(df.head())


# -----------------------------
# 1. BAR GRAPH
# -----------------------------

country_counts = df["country"].value_counts().head(10)

# Create a colormap based on count values
cmap_bar = plt.cm.viridis
colors_bar = cmap_bar(country_counts.values / float(max(country_counts.values)))

plt.figure(figsize=(10, 6))
bars = plt.bar(country_counts.index, country_counts.values, color=colors_bar)

plt.xlabel("Country")
plt.ylabel("Number of Earthquakes")
plt.title("Top 10 Countries by Number of Earthquakes")

plt.xticks(rotation=45)
plt.tight_layout()

# Add a color bar for the bar graph
sm_bar = ScalarMappable(cmap=cmap_bar, norm=Normalize(vmin=min(country_counts.values), vmax=max(country_counts.values)))
sm_bar.set_array(country_counts.values)
plt.colorbar(sm_bar, ax=plt.gca(), label='Number of Earthquakes')

plt.show()


# -----------------------------
# 2. HISTOGRAM
# -----------------------------

plt.figure(figsize=(10, 6))
n, bins, patches = plt.hist(df["magnitude"], bins=10, edgecolor='black')

# Create a colormap for the histogram bars based on their counts
cmap_hist = plt.cm.viridis
# Normalize counts to [0, 1] range for the colormap
n_norm = plt.Normalize(n.min(), n.max())
for i, patch in enumerate(patches):
    patch.set_facecolor(cmap_hist(n_norm(n[i])))

plt.xlabel("Magnitude")
plt.ylabel("Frequency")
plt.title("Distribution of Earthquake Magnitudes")

# Add a color bar for the histogram
sm_hist = ScalarMappable(cmap=cmap_hist, norm=Normalize(vmin=n.min(), vmax=n.max()))
sm_hist.set_array(n)
plt.colorbar(sm_hist, ax=plt.gca(), label='Frequency')

plt.show()


# -----------------------------
# 3. PIE CHART
# -----------------------------

plt.figure(figsize=(8, 8))
tsunami_counts = df["tsunami"].value_counts()

# Create a colormap for the pie chart slices based on their counts
cmap_pie = plt.cm.viridis
colors_pie = cmap_pie(tsunami_counts.values / float(max(tsunami_counts.values)))

plt.pie(
    tsunami_counts.values,
    labels=tsunami_counts.index,
    autopct="%1.1f%%",
    colors=colors_pie,
    startangle=90
)

plt.title("Earthquakes with Tsunami")

plt.show()


# -----------------------------
# 4. SCATTER PLOT
# -----------------------------

plt.figure(figsize=(10, 6))
scatter = plt.scatter(df["magnitude"], df["depth"], c=df["magnitude"], cmap='viridis', s=20, alpha=0.7)

plt.xlabel("Magnitude")
plt.ylabel("Depth")

plt.title("Magnitude vs Depth (Color by Magnitude)")
plt.colorbar(scatter, label='Magnitude')

plt.show()


# -----------------------------
# 5. LINE PLOT (modified to scatter with color by magnitude for colormap)
# -----------------------------

plt.figure(figsize=(12, 6))
# Using scatter plot to apply colormap based on magnitude for each point
scatter_line = plt.scatter(df.index, df["magnitude"], c=df["magnitude"], cmap='viridis', s=10, alpha=0.8)

plt.xlabel("Earthquake Number")
plt.ylabel("Magnitude")

plt.title("Earthquake Magnitude Trend (Color by Magnitude)")
plt.colorbar(scatter_line, label='Magnitude')

plt.show()
```
